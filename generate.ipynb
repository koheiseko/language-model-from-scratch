```python
import torch
from tokenizers import Tokenizer

from languagemodel.model import Transformer

tokenizer = Tokenizer.from_pretrained("NousResearch/Llama-2-7b-hf")

ckpt = torch.load("outputs/ckpt_14072026_14.pt", map_location="cuda")
model_args = ckpt["model_args"]
model = Transformer(**model_args)
model.load_state_dict(ckpt["model"])
```

```python
from languagemodel.generation import generate

prompt = "The capital of France is"
prompt_tokens = tokenizer.encode(prompt).ids

response = generate(
    model=model,
    max_new_tokens=10,
    eos_id=0,
    prompt_tokens=prompt_tokens,
    device="cuda",
    temperature=0.0,
).tolist()

print(tokenizer.decode(response))
```

```python
prompts = [
    "Alice picked up the red ball. Bob picked up the blue box. Alice was holding the",
    "Alice picked up the red ball. Bob picked up the blue box. Alice was holding the"
]

response = [
    tokenizer.decode(
        generate(
            model=model,
            max_new_tokens=10,
            eos_id=0,
            prompt_tokens=tokenizer.encode(prompt).ids,
            device="cuda",
            temperature=0.0,
        ).tolist()
    ) for prompt in prompts
]
```
